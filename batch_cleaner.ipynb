```python
from json_preprocessor.cleanJSON import cleanJSON
```

```python
import os
from pathlib import Path

def process_all_chunks(chunks_dir="chunks", output_dir="clean_chunks"):
    """
    Goes into each folder inside chunks_dir, finds the JSON with matching name,
    and runs cleanJSON on it.
    """
    chunks_path = Path(chunks_dir)
    
    # Check if chunks directory exists
    if not chunks_path.exists():
        print(f"Error: {chunks_dir} directory not found")
        return
    
    # Create output directory if it doesn't exist
    Path(output_dir).mkdir(exist_ok=True)
    
    # Iterate through each subdirectory in chunks
    for folder in chunks_path.iterdir():
        if folder.is_dir():
            folder_name = folder.name
            json_file = folder / f"{folder_name}.json"
            
            # Check if matching JSON exists
            if json_file.exists():
                print(f"Processing: {json_file}")
                try:
                    cleanJSON(str(json_file), output_dir)
                    print(f"✓ Successfully processed {folder_name}")
                except Exception as e:
                    print(f"✗ Error processing {folder_name}: {e}")
            else:
                print(f"⚠ Warning: {json_file} not found, skipping")

# Run it
process_all_chunks("chunks", "clean_chunks")
```
