```python
from selenium import webdriver
from bs4 import BeautifulSoup
import requests
import time

# Get page with Selenium
driver = webdriver.Chrome()
driver.get('YOUR_ORAN_URL')
time.sleep(5)

html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')

# Extract all download IDs
download_ids = []
for link in soup.find_all('a', href=True):
    if 'download?id=' in link['href']:
        doc_id = link['href'].split('id=')[-1]
        download_ids.append(doc_id)

driver.quit()

# Now download each file
for doc_id in download_ids:
    url = f"https://specifications.o-ran.org/download?id={doc_id}"
    response = requests.get(url)
    
    # Save file (adjust filename as needed)
    filename = f"oran_doc_{doc_id}.pdf"  # or extract from headers
    with open(filename, 'wb') as f:
        f.write(response.content)
    print(f"Downloaded: {filename}")
```
